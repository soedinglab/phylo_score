```python
import pickle
import numpy as np
```

```python
comap = pickle.load(open("data/eval_comap.pickle", "rb"))
phylo_score = pickle.load(open("data/eval_phyloscore.pickle", "rb"))
rscape = pickle.load(open("data/eval_rscape.pickle", "rb"))
```

```python
datasets = list(map(str.strip, open("3d.txt", "r")))
comap_datasets = [d for d in datasets if d not in ["RF00005", "RF03064"]]
```

```python
np.mean([comap[d, "corr"][1] for d in comap_datasets])
```

```python
np.mean([comap[d, "corrAPC"][1] for d in comap_datasets])
```
